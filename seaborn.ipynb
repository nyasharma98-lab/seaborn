```python
import seaborn as sns #seaborn is a python data visualization library based on matplotlib.
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
#relational plots ->scatter, line
sns.set_theme()
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker"
)
```

```python
sns.set_theme()
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="line"
)
```

```python
#scatter plot
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
sns.lineplot(
    data=tips,
    x="total_bill",
    y="tip"
)
```

```python
#line plot
flights = sns.load_dataset("flights")
flights.head()
```

```python
flights = sns.load_dataset("flights")

sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
```

```python
#categorical plots-> bar plot, box plot
print(tips.head())
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
#box plot
sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
#distribution plots-> histograms
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
```

```python
#matrix plot-> heatmaps-> plot rectangular data & use different colors fto represent values in a dataset.
flights.head()
```

```python
flights_pivot = flights.pivot(index="month", columns="year", values="passengers")
sns.heatmap(
    flights_pivot,
    cmap="coolwarm")
import matplotlib.pyplot as plt
plt.title("Passengers Heatmap")
```
